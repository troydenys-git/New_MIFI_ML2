### Метрика

В этом разделе вам необходимо посчитать метрику FID.

**Что такое FID?**

**FID (Fréchet Inception Distance)** — это метрика качества генеративных моделей для изображений, которая измеряет **расстояние между распределениями признаков реальных и сгенерированных изображений** в пространстве предобученной нейросети (обычно Inception-v3).

Чем **ниже FID**, тем **ближе** сгенерированные изображения к реальным — как по **качеству**, так и по **разнообразию**.

Формула FID основана на предположении, что признаки в этом пространстве приблизительно распределены как **многомерное нормальное распределение**:

$$
\text{FID} = \|\mu_r - \mu_g\|^2 + \mathrm{Tr}\left( \Sigma_r + \Sigma_g - 2\sqrt{\Sigma_r \Sigma_g} \right)
$$

где:
- $(\mu_r, \Sigma_r)$ — среднее и ковариационная матрица признаков **реальных** изображений,
- $(\mu_g, \Sigma_g)$ — то же для **сгенерированных** изображений,
- $\mathrm{Tr}(\cdot)$ — след матрицы.

> 🔹 FID = 0 означает полное совпадение распределений.  
> 🔹 Чем выше FID ↑ , тем качество или разнообразие генерации ниже ↓.

````python
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

# Define the device for training (GPU if available, else CPU)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Using device: {device}")

# Re-define VAE class (from prior cell)
class VAE(nn.Module):
    def __init__(self, latent_dim):
        super(VAE, self).__init__()
        self.latent_dim = latent_dim
        self.img_size = 28 * 28 # MNIST image size

        # Encoder: изображение -> mu, logvar
        self.encoder_fc1 = nn.Linear(self.img_size, 400)
        self.encoder_mu = nn.Linear(400, latent_dim)
        self.encoder_logvar = nn.Linear(400, latent_dim)

        # Decoder: z -> изображение
        self.decoder_fc1 = nn.Linear(latent_dim, 400)
        self.decoder_fc2 = nn.Linear(400, self.img_size)

    def encode(self, x):
        h = F.relu(self.encoder_fc1(x))
        mu = self.encoder_mu(h)
        logvar = self.encoder_logvar(h)
        return mu, logvar

    def reparameterize(self, mu, logvar):
        std = torch.exp(0.5 * logvar)
        eps = torch.randn_like(std)
        return mu + eps * std

    def decode(self, z):
        h = F.relu(self.decoder_fc1(z))
        return torch.sigmoid(self.decoder_fc2(h))

    def forward(self, x):
        x = x.view(-1, self.img_size)
        mu, logvar = self.encode(x)
        z = self.reparameterize(mu, logvar)
        recon_x = self.decode(z)
        return recon_x, mu, logvar

# Re-define vae_loss function (from prior cell)
def vae_loss(recon_x, x, mu, logvar):
    # Denormalize x from [-1, 1] to [0, 1] for BCE Loss target
    x_denorm = (x + 1) / 2
    # Reconstruction loss: BCE (since output is sigmoid)
    recon_loss = F.binary_cross_entropy(recon_x, x_denorm.view(-1, 28 * 28), reduction='sum')

    # KL divergence: D_KL(q(z|x) || p(z))
    # 0.5 * sum(1 + log(sigma^2) - mu^2 - sigma^2)
    kl_loss = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp())

    return recon_loss + kl_loss

# Re-define training parameters (from prior cell)
latent_dim = 20
epochs = 10
lr = 1e-3
img_size = 28
channels = 1

# Re-define train_loader (from prior cell, assuming batch_size)
batch_size = 128
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.5,), (0.5,))
])
train_dataset = datasets.MNIST(root='./data', train=True, download=True, transform=transform)
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)


model = VAE(latent_dim).to(device)
optimizer = optim.Adam(model.parameters(), lr=lr)

# Training loop
print("Starting VAE training...")
for epoch in range(1, epochs + 1):
    model.train()
    train_loss = 0
    for batch_idx, (data, _) in enumerate(train_loader):
        data = data.to(device)
        optimizer.zero_grad()
        recon_batch, mu, logvar = model(data)
        loss = vae_loss(recon_batch, data, mu, logvar)
        loss.backward()
        train_loss += loss.item()
        optimizer.step()

    avg_train_loss = train_loss / len(train_loader.dataset)
    print(f'Epoch: {epoch} Average loss: {avg_train_loss:.4f}')

print("VAE training complete.")
````

````python
import os
import subprocess
from PIL import Image
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt

# Определяем устройство для обучения (GPU, если доступно, иначе CPU)
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Используемое устройство: {device}")

# Переопределяем класс VAE для безусловной генерации
class VAE(nn.Module):
    def __init__(self, latent_dim):
        super(VAE, self).__init__()
        self.latent_dim = latent_dim
        self.img_size_flat = 28 * 28 # Размер изображения MNIST (28x28) в "плоском" виде

        # Кодировщик: преобразует изображение в параметры скрытого пространства (mu, logvar)
        self.encoder_fc1 = nn.Linear(self.img_size_flat, 400)
        self.encoder_mu = nn.Linear(400, latent_dim) # Выход для среднего (mu) скрытого пространства
        self.encoder_logvar = nn.Linear(400, latent_dim) # Выход для логарифма дисперсии (logvar)

        # Декодировщик: преобразует вектор скрытого пространства (z) обратно в изображение
        self.decoder_fc1 = nn.Linear(latent_dim, 400)
        self.decoder_fc2 = nn.Linear(400, self.img_size_flat)

    def encode(self, x):
        # Пропускаем входное изображение через полносвязные слои энкодера
        h = F.relu(self.encoder_fc1(x))
        mu = self.encoder_mu(h)
        logvar = self.encoder_logvar(h)
        return mu, logvar

    def reparameterize(self, mu, logvar):
        # Применяем трюк репараметризации для получения выборки из скрытого пространства
        std = torch.exp(0.5 * logvar) # Вычисляем стандартное отклонение из логарифма дисперсии
        eps = torch.randn_like(std) # Генерируем случайный шум из стандартного нормального распределения
        return mu + eps * std # Возвращаем выборку z

    def decode(self, z):
        # Пропускаем вектор скрытого пространства z через полносвязные слои декодера
        h = F.relu(self.decoder_fc1(z))
        return torch.sigmoid(self.decoder_fc2(h)) # Используем сигмоиду для получения пикселей в диапазоне [0, 1]

    def forward(self, x):
        # Прямой проход VAE: кодирование, репараметризация, декодирование
        x = x.view(-1, self.img_size_flat) # "Разворачиваем" изображение в одномерный вектор
        mu, logvar = self.encode(x)
        z = self.reparameterize(mu, logvar)
        recon_x = self.decode(z)
        return recon_x, mu, logvar

# Переопределяем функцию потерь VAE
def vae_loss(recon_x, x, mu, logvar):
    # Целевое изображение для BCE должно быть в диапазоне [0, 1].
    # Мы нормализовали данные до [-1, 1], поэтому денормализуем обратно в [0, 1].
    x_denorm = (x + 1) / 2
    # Потери реконструкции: бинарная кросс-энтропия (BCE), так как выход сигмоида
    recon_loss = F.binary_cross_entropy(recon_x, x_denorm.view(-1, 28 * 28), reduction='sum')

    # KL-дивергенция: D_KL(q(z|x) || p(z))
    # Формула: 0.5 * sum(1 + log(sigma^2) - mu^2 - sigma^2)
    kl_loss = -0.5 * torch.sum(1 + logvar - mu.pow(2) - logvar.exp())

    return recon_loss + kl_loss

# Переопределяем параметры обучения
latent_dim = 20 # Размерность скрытого пространства
epochs = 10 # Количество эпох обучения
lr = 1e-3 # Скорость обучения
img_size = 28 # Размер изображения (сторона)
channels = 1 # Количество каналов (для MNIST 1, так как черно-белые изображения)

# Переинициализируем датасеты и загрузчики, чтобы они были доступны
batch_size = 128 # Размер батча
transform = transforms.Compose([
    transforms.ToTensor(), # Преобразует изображение в тензор PyTorch
    transforms.Normalize((0.5,), (0.5,)) # Нормализует пиксели до диапазона [-1, 1]
])
train_dataset = datasets.MNIST(root='./data', train=True, download=True, transform=transform)
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
test_dataset = datasets.MNIST(root='./data', train=False, download=True, transform=transform)

# Инициализируем и обучаем модель
model = VAE(latent_dim).to(device)
optimizer = optim.Adam(model.parameters(), lr=lr)

print("Начинаем обучение VAE...")
for epoch in range(1, epochs + 1):
    model.train() # Устанавливаем модель в режим обучения
    train_loss = 0
    for batch_idx, (data, _) in enumerate(train_loader):
        data = data.to(device)
        optimizer.zero_grad() # Обнуляем градиенты
        recon_batch, mu, logvar = model(data) # Прямой проход модели
        loss = vae_loss(recon_batch, data, mu, logvar) # Вычисляем потери
        loss.backward() # Обратное распространение ошибки
        train_loss += loss.item()
        optimizer.step() # Обновляем веса модели

    avg_train_loss = train_loss / len(train_loader.dataset)
    print(f'Эпоха: {epoch} Средние потери: {avg_train_loss:.4f}')

print("Обучение VAE завершено.")

# Создаем и заполняем директорию 'mnist_vae_real' для реальных изображений
real_images_dir = 'mnist_vae_real'
os.makedirs(real_images_dir, exist_ok=True)
num_fid_images = 10000
print(f"Пересохраняем {num_fid_images} реальных тестовых изображений MNIST в '{real_images_dir}' для расчета FID...")
for i in range(num_fid_images):
    img, _ = test_dataset[i] # Получаем изображение и метку
    img_np = (img.squeeze().numpy() * 0.5) + 0.5 # Денормализуем из [-1, 1] в [0, 1]
    img_pil = Image.fromarray((img_np * 255).astype(np.uint8)) # Преобразуем в PIL Image
    img_pil.save(os.path.join(real_images_dir, f'img_{i:05d}.png'))
print(f"Успешно пересохранено {len(os.listdir(real_images_dir))} реальных изображений.")

# Создаем и заполняем директорию 'mnist_vae_fake' для сгенерированных изображений
fake_images_dir = 'mnist_vae_fake'
os.makedirs(fake_images_dir, exist_ok=True)

# Генерируем и сохраняем поддельные изображения
print(f"Генерируем и сохраняем {num_fid_images} поддельных изображений в '{fake_images_dir}' для расчета FID...")
model.eval() # Устанавливаем модель в режим оценки
with torch.no_grad(): # Отключаем расчет градиентов
    for i in range(num_fid_images):
        sample = torch.randn(1, latent_dim).to(device) # Выбираем случайный вектор из скрытого пространства
        generated_img = model.decode(sample).cpu() # Декодируем для получения изображения
        img_np = generated_img.view(img_size, img_size).numpy() # Изменяем форму и преобразуем в numpy
        img_pil = Image.fromarray((img_np * 255).astype(np.uint8)) # Преобразуем в PIL Image
        img_pil.save(os.path.join(fake_images_dir, f'img_{i:05d}.png'))
print(f"Успешно сгенерировано и сохранено {len(os.listdir(fake_images_dir))} поддельных изображений.")

# Убеждаемся, что pytorch-fid установлен
print("\nУбеждаемся, что pytorch-fid установлен...")
!pip install pytorch-fid
print("Установка pytorch-fid подтверждена.")

# Проверяем директории перед запуском FID
print(f"\nПроверяем директорию: {real_images_dir}")
if os.path.isdir(real_images_dir):
    real_files = [f for f in os.listdir(real_images_dir) if os.path.isfile(os.path.join(real_images_dir, f))]
    print(f"Найдено {len(real_files)} файлов в {real_images_dir}")
else:
    print(f"Директория {real_images_dir} не существует.")

print(f"Проверяем директорию: {fake_images_dir}")
if os.path.isdir(fake_images_dir):
    fake_files = [f for f in os.listdir(fake_images_dir) if os.path.isfile(os.path.join(fake_images_dir, f))]
    print(f"Найдено {len(fake_files)} файлов в {fake_images_dir}")
else:
    print(f"Директория {fake_images_dir} не существует.")

# Вычисляем показатель FID
print(f"\nВычисляем показатель FID между '{real_images_dir}' и '{fake_images_dir}'...")
fid_command = ['python', '-m', 'pytorch_fid', real_images_dir, fake_images_dir, '--device', 'cpu'] # Используем 'cpu', так как VAE обучался на CPU

fid_value = None
try:
    print("Запуск команды pytorch_fid...")
    result = subprocess.run(fid_command, capture_output=True, text=True, check=True)
    print("Команда pytorch_fid завершена.")
    print("Вывод расчета FID:")
    print(result.stdout)
    if result.stderr:
        print("Ошибка расчета FID (если есть, обычно для некритических предупреждений):\n")
        print(result.stderr)

    # Поиск FID значения в выводе
    fid_line = next((line for line in result.stdout.splitlines() if line.startswith('FID:')), None)
    if fid_line:
        try:
            fid_value = float(fid_line.split(':')[1].strip())
            print(f"\nОбнаруженное FID значение: {fid_value:.4f}")
        except ValueError:
            print(f"Не удалось распарсить FID значение из строки: {fid_line}")
    else:
        print("FID значение не найдено в выводе.")

except subprocess.CalledProcessError as e:
    print(f"Расчет FID завершился с кодом выхода {e.returncode}.")
    print(f"Команда: {' '.join(e.cmd)}")
    print("Stderr:")
    print(e.stderr)
    print("Stdout:")
    print(e.stdout)
except FileNotFoundError:
    print("Ошибка: Команда 'python' не найдена. Убедитесь, что Python находится в вашем PATH.")

print("\nРасчет FID завершен.")

# Добавляем итоговое резюме, чтобы значение было четко зафиксировано
if fid_value is not None:
    print(f"\n--- Итоговый FID для безусловного VAE: {fid_value:.4f} ---")
else:
    print("\n--- Не удалось получить итоговый FID для безусловного VAE ---")
````
