## Параметры расчета
- **Количество групп**: 12
- **Минимальный детектируемый эффект**: 5% (0.05)
- **Мощность теста**: 80% (β = 0.2)
- **Уровень значимости**: 5% (α = 0.05)
- **Тип теста**: Двусторонний

## Методология
Для решения задачи нам нужно рассчитать минимальный размер выборки на группу для MVT-теста с 12 группами, чтобы детектировать эффект 5% с мощностью 80% и уровнем значимости 5%.
Мы будем использовать формулу для расчета размера выборки в A/B-тестах (или MVT) для пропорций (конверсии). 
### Используемая формула
Для расчета размера выборки на группу используется стандартная формула для сравнения пропорций:


$n = \frac{(Z_{1-α/2} + Z_{1-β})² × (p₁(1-p₁) + p₂(1-p₂))}{(p₁ - p₂)²}$

где:
- $p₁$ - базовая конверсия (контроль)
- $p₂ = p₁ + эффект$ - конверсия с эффектом
- $Z_{1-\frac{α}{2}}$ - Z-score для уровня значимости
- $Z_{1-β}$ - Z-score для мощности

### Расчетные значения
- **$Z_{1-α/2}$** (α=0.05, двусторонний): 1.9695
- **$Z_{1-β}$** (β=0.20): 0.8416

1. В чем основное отличие MVT от A/B-тестирования?
Основное отличие между A/B и MVT-тестами заключается в количестве одновременно проверяемых переменных и сложности их взаимодействий. Цель MVT-теста - определить не только лучшие комбинации проверяемых переменных, но и взаимодействие факторов.

2. Какие метрики обычно используются для оценки эффективности MVT?
    - Бизнес-метрики: коэффициент конверсии, средний чек, CTR (кликабельность элементов), доход на посетителя.
    - Специфические метрики взаимодействий элементов: 
        вклад элемента - показывает, какая переменная (например, заголовок или изображение) оказала наибольшее влияние на итоговый результат;
        эффект взаимодействия - оценка того, как два элемента работают «в связке», например, синяя кнопка может работать плохо сама по себе, но в сочетании с конкретным заголовком давать лучший результат;
        уровень достоверности - статистическая значимость для каждой выигрышной комбинации. 
    - Поведенческие метрики: 
        показатель отказов - как изменения влияют на мгновенный уход пользователей со страницы;
        глубина просмотра и время на странице - метрики вовлеченности, показывающие интерес к контенту;
        коэффициент микроконверсий - промежуточные шаги (например, добавление в корзину без покупки), помогающие увидеть прогресс даже при малом количестве финальных продаж.
    - Метрики стабильности и качества:
        уровень ошибок (Error Rate) - мониторинг технических сбоев в разных комбинациях элементов, чтобы исключить влияние багов на результаты теста;
        RMSE (Root Mean Square Error) - используется для оценки точности моделей предсказания поведения пользователей в рамках теста.

3. Как определить количество групп в MVT?
Количество групп зависит от количества комбинаций тестируемых элементов. Максимальное количество групп определяется как произведение количества вариантов для каждого тестируемого элемента, включая контрольный вариант. Например для тестирования  3х заголовков и 2х вариантов изображений необходимо 3*2 = 6 групп. При недостатке ресурсов, проводится дробный факторный дизайн (Fractional Factorial) - из общего количества групп отбираются наиболее значимые.

4. Какие риски связаны с проведением MVT?
    - Недостаток трафика. При большом количестве комбинаций есть риск не набрать необходимое количество трафика и выборка пользователей окажется нерепрезентативной.
    - Возможность ложных срабатываний. При увеличении вариантов комбинаций, возрастает риск случайных действий пользователей. Многократное тестирование значимости всегда увеличивает частоту ложноположительных результатов
    - Данные ненормальные/сильно скошены (например, доход на пользователя). Реальные данные обычно отличаются от нормальных, есть риск выбрать неверный метод оценки.
    - Когда аудитория разбивается более чем на 2 подгруппы, вероятность получения ошибки I рода (FP) растет экспоненциально. 


5. Как интерпретировать взаимодействие факторов в MVT?
Интерпретация взаимодействия факторов в MVT это понимание того, как комбинация изменений элементов влияет на целевую метрику. Необходимо оценить именно взаимодействие элементов теста, напирмер, может быть положительное взаимодействие, если комбинация элементов работает лучше, чем каждый элемент по отдельности, либо отрицательное взаимодействие, когда элементы вместе работают хуже, чем по отдельности.
Инструменты интерпретации взаимодейтсвия факторов в MVT тесте:
    - Специализированные платформы оптимизации: Adobe Target, VWO (Visual Website Optimizer), Optimizely;
    - Статистические методы и модели: дисперсионный анализ (ANOVA),регрессионный анализ, байесовские модели;
    - Инструменты визуализации данных: графики взаимодействия, тепловые карты, матрицы диаграмм рассеяния.
