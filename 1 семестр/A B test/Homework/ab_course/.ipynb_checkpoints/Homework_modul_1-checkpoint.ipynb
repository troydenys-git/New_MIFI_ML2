# Статистический анализ двух бинарных выборок

## 📊 Условие задачи

### Исходные данные:
- **Две бинарные выборки** с вероятностями успеха:
  - $p_1 = 0.35$ (первая выборка)
  - $p_2 = 0.40$ (вторая выборка)
- **Уровень значимости**: $\alpha = 0.05$ (p-value)
- **Минимальный детектируемый эффект (MDE)**: $\delta = |p_2 - p_1| = 0.05$

### Поставленные задачи:

#### 1. 📈 Определение минимального размера выборки
**Цель**: Найти минимальное количество элементов $n$ в каждой выборке, необходимое для детекции MDE с заданной мощностью теста ($1 - \beta = 0.8$).

**Формула для расчета**:
$$
n = \frac{(z_{1-\alpha/2} \cdot \sqrt{2 \cdot \bar{p} \cdot (1-\bar{p})} + z_{1-\beta} \cdot \sqrt{p_1(1-p_1) + p_2(1-p_2)})^2}{(p_1 - p_2)^2}
$$

где:
- $\bar{p} = \frac{p_1 + p_2}{2}$ - объединенная пропорция
- $z_{1-\alpha/2}$ - квантиль нормального распределения для двустороннего теста
- $z_{1-\beta}$ - квантиль нормального распределения для мощности теста

#### 2. 🔄 Преобразование распределений через ЦПТ
**Цель**: Используя Центральную Предельную Теорему (ЦПТ), преобразовать биномиальные распределения к нормальным.

**Формулы преобразования**:
- Для выборки размера $n$ с вероятностью $p$:
  $$
  \text{Среднее: } \mu = p
  $$
  $$
  \text{Стандартное отклонение: } \sigma = \sqrt{\frac{p(1-p)}{n}}
  $$
- Получаем нормальные распределения:
  - Первая выборка: $N(\mu_1, \sigma_1)$
  - Вторая выборка: $N(\mu_2, \sigma_2)$

#### 3. 📊 Визуализация распределений и ошибок
**Требуется построить два графика**:

**График 1**: Двусторонняя альтернатива
- Распределения обеих выборок
- Критические области для ошибки I рода ($\alpha$)
- Область ошибки II рода ($\beta$)

**График 2**: Односторонняя альтернатива ($p_1 > p_2$)
- Распределения обеих выборок  
- Односторонние критические области
- Соответствующие области ошибок

**Обозначения на графиках**:
- 🟦 **Синяя область** - Ошибка I рода (вероятность $\alpha$)
- 🟥 **Красная область** - Ошибка II рода (вероятность $\beta$)
- 📍 **Вертикальные линии** - Критические значения и средние

#### 4. 📐 Расчет доверительных интервалов
**Цель**: Построить $95\%$ доверительные интервалы для средних значений обеих выборок.

**Формула доверительного интервала**:
$$
CI = \hat{p} \pm z_{1-\alpha/2} \cdot \sqrt{\frac{\hat{p}(1-\hat{p})}{n}}
$$

где:
- $\hat{p}$ - выборочная пропорция
- $z_{1-\alpha/2}$ - квантиль стандартного нормального распределения

### 🎯 Ожидаемые результаты:

1. **Числовой результат**: Минимальный размер выборки $n$
2. **Параметры распределений**: $\mu_1, \sigma_1, \mu_2, \sigma_2$
3. **Визуализация**: Два графика с областями статистических ошибок
4. **Интервальные оценки**: Доверительные интервалы для $p_1$ и $p_2$

### 📈 Статистические гипотезы:

**Для двустороннего теста**:
$$
H_0: p_1 = p_2 \\
H_1: p_1 \neq p_2
$$

**Для одностороннего теста**:
$$
H_0: p_1 \leq p_2 \\
H_1: p_1 > p_2
$$

```python
def plot_two_sided_test(mu1, mu2, se, alpha):

    # 1. Настройка оси X

    x = np.linspace(min(mu1, mu2) - 4*se, max(mu1, mu2) + 4*se, 1000)

    # 2. Создание распределений 
    dist_h0 = stats.norm(loc=mu1, scale=se) # Нулевая гипотеза
    dist_h1 = stats.norm(loc=mu2, scale=se) # Альтернативная гипотеза

    # 3. Расчет критических значений
    # alpha делится пополам на два хвоста
    crit_left = dist_h0.ppf(alpha / 2)
    crit_right = dist_h0.ppf(1 - alpha / 2)

    # 4. Построение графиков
    plt.figure(figsize=(9, 4))

    # Линии плотности вероятности (PDF)
    plt.plot(x, dist_h0.pdf(x), color='black', label='$H_0$ (Нулевая гипотеза)', linewidth=2)
    plt.plot(x, dist_h1.pdf(x), color='orange', linestyle='--', label='$H_1$ (Альтернатива)', linewidth=2)

    # 5. Заливка областей

    # Ошибка первого рода

    plt.fill_between(x, dist_h0.pdf(x), where=(x <= crit_left), 
                     color='blue', alpha=0.5, label='Ошибка 1 рода ($\\alpha/2$)')

    plt.fill_between(x, dist_h0.pdf(x), where=(x >= crit_right), 
                     color='blue', alpha=0.5)

    # Ошибка второго рода

    plt.fill_between(x, dist_h1.pdf(x), where=((x >= crit_left) & (x <= crit_right)), 
                     color='red', alpha=0.3, label='Ошибка 2 рода ($\\beta$)')

    # 6. Вертикальные линии
    plt.axvline(crit_left, color='red', linestyle=':', linewidth=1.5)
    plt.axvline(crit_right, color='red', linestyle=':', linewidth=1.5, label=f'Крит. значения (Z)')
    plt.axvline(mu1, color='grey', linestyle='-', alpha=0.5, ymax=0.95)
    plt.axvline(mu2, color='grey', linestyle='--', alpha=0.5, ymax=0.95)

    # Подписи и легенда
    plt.title(f'Двусторонняя альтернатива: H_0: p1=p2 vs H_1: p1≠p2', fontsize=14)
    plt.xlabel('Значения статистики', fontsize=12)
    plt.ylabel('Плотность вероятности', fontsize=12)
    plt.legend(loc='upper left', frameon=True)
    plt.grid(True, alpha=0.3)
    
    # Текстовые аннотации для наглядности (опционально)
    plt.text(mu1, dist_h0.pdf(mu1)*0.4, '$\mu_0$', ha='center', fontsize=12, fontweight='bold')
    plt.text(mu2, dist_h1.pdf(mu2)*0.4, '$\mu_1$', ha='center', fontsize=12, fontweight='bold')

    plt.show()
```

```python
def plot_one_sided_test(mu0, mu1, se, alpha):

    # 1. Настройка оси X
    x = np.linspace(min(mu0, mu1) - 4*se, max(mu0, mu1) + 4*se, 1000)

    # 2. Создание распределений 
    dist_h0 = stats.norm(loc=mu0, scale=se) # Нулевая гипотеза
    dist_h1 = stats.norm(loc=mu1, scale=se) # Альтернативная гипотеза

    # 3. Расчет критического значения

    crit_val = dist_h0.ppf(1 - alpha)

    # 4. Построение графиков
    plt.figure(figsize=(10, 5))

    # Линии плотности вероятности (PDF)
    plt.plot(x, dist_h0.pdf(x), color='black', label='$H_0$ (Нулевая гипотеза)', linewidth=2)
    plt.plot(x, dist_h1.pdf(x), color='orange', linestyle='--', label='$H_1$ (Альтернатива)', linewidth=2)

    # 5. Заливка областей

    # Ошибка первого рода 
    plt.fill_between(x, dist_h0.pdf(x), where=(x <= crit_val), 
                     color='blue', alpha=0.5, label='Ошибка 1 рода ($\\alpha$)')

    # Ошибка второго рода 

    plt.fill_between(x, dist_h1.pdf(x), where=(x >= crit_val), 
                     color='red', alpha=0.3, label='Ошибка 2 рода ($\\beta$)')

    # 6. Вертикальные линии
    plt.axvline(crit_val, color='red', linestyle=':', linewidth=1.5, 
                label=f'Крит. значение (Z)')
    plt.axvline(mu0, color='grey', linestyle='-', alpha=0.5, ymax=0.95)
    plt.axvline(mu1, color='grey', linestyle='--', alpha=0.5, ymax=0.95)

    # Подписи и легенда
    plt.title(f'Односторонняя альтернатива: H_0: p1 <= p2 vs H_1: p1 > p2', fontsize=14)
    plt.xlabel('Значения статистики', fontsize=12)
    plt.ylabel('Плотность вероятности', fontsize=12)
    
    #plt.ylim(bottom=0)
    
    plt.legend(loc='upper left', frameon=True)
    plt.grid(True, alpha=0.3)
    
    # Текстовые аннотации
    plt.text(mu0, dist_h0.pdf(mu0)*0.4, '$\mu_0$', ha='center', fontsize=12, fontweight='bold')
    plt.text(mu1, dist_h1.pdf(mu1)*0.4, '$\mu_1$', ha='center', fontsize=12, fontweight='bold')

    plt.show()
```

```python
# Ваш код

from scipy.stats import norm
import numpy as np
import math
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import scipy.stats as stats
# Исходные данные

p1 = 0.35
p2 = 0.4
alpha = 0.05
power = 0.8

# 1. Определение минимального размера выборки

def sample_size(p1, p2, alpha, power):
   
   # Квантили нормального распределения 
    z_alpha = norm.ppf(1-alpha/2)
    z_beta = norm.ppf(power)
    
    # Объединенная пропорция
    p_bar = (p1 + p2)/2
    
    t_alpha = z_alpha*((2*p_bar*(1-p_bar))**0.5)
    t_beta = z_beta*np.sqrt(p1*(1-p1) + p2*(1-p2))
    
    # Расчет выборки
    n = ((t_alpha + t_beta)**2)/((p1 - p2)**2) 
    return n

n_sample = math.ceil(sample_size(p1, p2, alpha, power))

# 2. Преобразование распределений через ЦПТ

mu1 = p1
mu2 = p2
# Расчет стандартного отклонения
sigma1 = np.sqrt((p1*(1-p1))/n_sample)
sigma2 = np.sqrt((p2*(1-p2))/n_sample)

# 4. Расчет доверительных интервалов
# Квантиль
z = stats.norm.ppf(1 - alpha/2)
# Стандартные ошибки
se1 = np.sqrt(p1 * (1 - p1) / n_sample)
se2 = np.sqrt(p2 * (1 - p2) / n_sample)
# Интервалы
ci1_lower = p1 - z * se1
ci1_upper = p1 + z * se1
ci2_lower = p2 - z * se2
ci2_upper = p2 + z * se2


# ОТВЕТ

print('='*80)
print('\033[1m' + 'АНАЛИЗ ДВУХ БИНАРНЫХ ВЫБОРОК (С РАВНЫМИ РАЗМЕРАМИ)')
print('='*80 + '\033[0m')
print()
print(f'Вероятность в группе 1: {p1}')
print(f'Вероятность в группе 2: {p2}')
print(f'Уровень значимости: {alpha}')
print('\033[1m' + f'Мощность теста: {power}')
print('-'*80 + '\033[0m')
print('1. МИНИМАЛЬНЫЙ РАЗМЕР ВЫБОРКИ (РАВНЫЕ ГРУППЫ):')
print()
print(f'n₁ = n₂ = {n_sample} элементов в каждой группе')
print()
print('2. ПРЕОБРАЗОВАНИЕ ЧЕРЕЗ ЦПТ:')
print()
print('РАВНЫЕ ВЫБОРКИ (n=1471):')
print()
print(f'Группа 1: N(μ = {mu1:.4f}, σ = {sigma1:.4f})')
print(f'Группа 2: N(μ = {mu2:.4f}, σ = {sigma2:.4f})')
print()
print('3. Двусторонняя альтернатива')
print()
plot_two_sided_test(mu1=0.35, mu2=0.4, se=0.2, alpha=0.05)
print()
print('Односторонняя альтернатива')
plot_right_sided_test(mu0=0.35, mu1=0.4, se=0.2, alpha=0.05)
print()
print('4. ДОВЕРИТЕЛЬНЫЕ ИНТРЕВАЛЫ (уровень доверия 0.95):')
print()
print('РАВНЫЕ ВЫБОРКИ:')
print()
print(f'Группа 1: [{ci1_lower:.4f}, {ci1_upper:.4f}], (ширина: ({(ci1_upper-ci1_lower):.4f})')
print(f'Группа 2: [{ci2_lower:.4f}, {ci2_upper:.4f}], (ширина:({(ci2_upper-ci2_lower):.4f})')
```
