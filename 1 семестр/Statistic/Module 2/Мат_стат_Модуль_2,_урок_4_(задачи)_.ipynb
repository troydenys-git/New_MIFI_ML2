```python
## Импорт библиотек
import numpy as np
from   scipy.stats import randint
from   scipy.stats import bernoulli
from   scipy.stats import binom
from   scipy.stats import geom
from   scipy.stats import nbinom
from   scipy.stats import hypergeom
from   scipy.stats import poisson
```

**Задача 4.1**

Смоделируйте 15000 случайных величин из дискретного равномерного распределения на множестве ${3,4,\dots,12}$. Найдите выборочное математическое ожидание и дисперсию, округлите до трёх знаков после запятой. Введите разницу между теоретическим и выборочным матожиданием.

```python
#Заполните пропуски "___"

a = 3
b = 13
size = 15000

# Генерация данных
data = randint.rvs(low=a, high=b, size=size, random_state=42)

# Выборочные характеристики
sample_mean = round(np.mean(data), 3)
sample_var = round(np.var(data), 3)

# Теоретические значения
theor_mean = randint.mean(a, b)
diff_mean = theor_mean - sample_mean

print(f"Выборочное среднее: {sample_mean}")
print(f"Теоретическое среднее: {theor_mean}")
print(f"Разница: {diff_mean}")
```

**Задача 4.2**

Проведите $10000$ экспериментов с вероятностью успеха $p = 0.42$. Посчитайте долю успешных исходов. На сколько она отличается от заданного $p$? Ответ округлите до трёх знаков.

```python
#Заполните пропуски "___"

p = 0.42
n_trials = 10000

# Генерация данных
results = bernoulli.rvs(p, loc=0, size=n_trials, random_state=None)

# Подсчёт частоты успехов
success_rate = np.mean(results)

# Разница
diff = success_rate - p

print(f"Доля успехов: {success_rate}")
print(f"Разница с p: {diff}")
```

**Задача 4.3**

Подбросьте монету $25$ раз $(p=0.5)$, повторив процедуру $50000$ раз. Найдите долю испытаний, где число успехов лежит в интервале от $12$ до $18$ включительно. Ответ округлите до трёх знаков.

```python
#Заполните пропуски "___"

n = 25
p = 0.5
n_simulations = 50000

# Симуляция
simulations = binom.rvs(n, p, loc=0, size=n_simulations, random_state=None)

# Фильтрация
in_range = [12,18]
probability = binom.cdf(in_range[1], n, p) - binom.cdf(in_range[0] - 1, n, p)

print(f"Вероятность в диапазоне: {probability:.3f}")
```

**Задача 4.4**

Смоделируйте $8000$ независимых испытаний с $p=0.23$. Найдите среднее число попыток до первого успеха. Насколько оно отличается от теоретического значения $1/p$? Ответ округлите до трёх знаков.

```python
#Заполните пропуски "___"

p = 0.23
n_simulations = 8000

# Генерация данных
trials_until_success = geom.rvs(p, loc=0, size=n_simulations, random_state=None)

# Среднее
mean_trials = np.mean(trials_until_success)

# Теоретическое значение
theor_mean = geom.mean(p)

# Разница
diff = mean_trials - theor_mean

print(f"Среднее число попыток: {mean_trials}")
print(f"Разница с теорией: {diff}")
```

**Задача 4.5**

Смоделируйте $5000$ ситуаций, где нужно получить $7$ успехов с $p=0.35$. Какова доля случаев, в которых потребуется больше $20$ неудач? Ответ округлите до трёх знаков.

```python
#Заполните пропуски "___"

r = 7
p = 0.35
n_simulations = 5000

# Генерация числа неудач до r-го успеха
failures = nbinom.rvs(r, p, loc=0, size=n_simulations, random_state=None)

# Подсчёт доли > 20
prob_more_than_20 = nbinom.pmf(20, r, p)

print(f"Доля > 20 неудач: {prob_more_than_20}")
failures
```

**Задача 4.6**

В урне $35$ белых и #15# чёрных шаров. Наугад вынимают $7$ шаров без возвращения. Какова вероятность, что среди них будет не менее $4$ белых? Ответ округлите до трёх знаков.



```python
#Заполните пропуски "___"

M = 50   # всего шаров
n = 35   # белых
N = 7   # выбираем
n_simulations = 7

# Симуляции
draws = hypergeom.rvs(M, n, N, loc=0)

# Доля >= 4
prob_at_least_4 = hypergeom.sf(4 - 1, M, n, N)

print(f"Вероятность ≥ 4 белых: {prob_at_least_4}")
```

**Задача 4.7**
