```python
# -*- coding: utf-8 -*-
import json
import pandas as pd

# Укажите путь к вашему JSON файлу
json_file = 'C:\IDE\Стажировка\Настя.json'  # Измените на имя вашего файла

# Укажите имя выходного Excel файла
excel_file = 'График_Работ_Питомника_2026.xlsx'

# Читаем JSON
with open(json_file, 'r', encoding='utf-8') as f:
    data = json.load(f)

# Создаем Excel с несколькими листами
with pd.ExcelWriter(excel_file, engine='openpyxl') as writer:
    for sheet in data['sheets']:
        # Берем название листа
        sheet_name = sheet['name']
        rows = sheet['rows']
        
        # Первая строка - заголовки, остальные - данные
        headers = rows[0]
        data_rows = rows[1:]
        
        # Создаем таблицу
        df = pd.DataFrame(data_rows, columns=headers)
        
        # Записываем в Excel
        df.to_excel(writer, sheet_name=sheet_name, index=False)
        print(f'Добавлен лист: {sheet_name}')

print(f'\nГотово! Файл сохранен: {excel_file}')

C:\IDE\График_Работ_Питомника_2026.xlsx
```
