```python
!pip install -q kagglehub tensorflow
import kagglehub, tensorflow as tf

# 1. Download model
model_path = kagglehub.model_download("wafaa/cats-vs-dogs-finetuned-mobilenetv2")

# 2. Load model
model = tf.keras.models.load_model(model_path)

# 3. Prepare input (expects float32 tensor [224,224,3], scaled 0–1)
img = tf.keras.utils.load_img("/path/to/image.jpg", target_size=(224, 224))
img = tf.keras.utils.img_to_array(img) / 255.0
pred = model.predict(tf.expand_dims(img, 0))[0][0]
print("Dog 🐶" if pred >= 0.5 else "Cat 🐱")

```
