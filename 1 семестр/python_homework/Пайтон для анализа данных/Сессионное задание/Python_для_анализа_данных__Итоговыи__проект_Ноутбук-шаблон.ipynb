1. Начнем с простого - с признака **"Образование и ВУЗ"**. Его текущий формат это: **<Уровень образования год выпуска ВУЗ специальность...>**. Например:
* Высшее образование 2016 Московский авиационный институт (национальный исследовательский университет)...
* Неоконченное высшее образование 2000  Балтийская государственная академия рыбопромыслового флота…
Нас будет интересовать только уровень образования.

Создайте с помощью функции-преобразования новый признак **"Образование"**, который должен иметь 4 категории: "высшее", "неоконченное высшее", "среднее специальное" и "среднее".

Выполните преобразование, ответьте на контрольные вопросы и удалите признак "Образование и ВУЗ".

Совет: обратите внимание на структуру текста в столбце **"Образование и ВУЗ"**. Гарантируется, что текущий уровень образования соискателя всегда находится в первых 2ух слов и начинается с заглавной буквы. Воспользуйтесь этим.

*Совет: проверяйте полученные категории, например, с помощью метода unique()*


```python
#ваш код здесь
# СОздадим копию датасета для преобразования
HeadHunter_data_feat_eng = HeadHunter_data.copy()

educ_train = HeadHunter_data_feat_eng['Образование и ВУЗ']

cat_education = ["высшее", "неоконченное высшее", "среднее специальное","среднее"]

def transform_education(education):
    
    education = education.split()[:2]
    if education[0] == 'Высшее':
        return "высшее"
    elif education[1] == 'специальное':
        return "среднее специальное"
    elif education[0] == 'Среднее':
        return "среднее"
    elif education[0] == 'Неоконченное':
        return "неоконченное высшее"
    return education

educ_train = educ_train.apply(transform_education)
HeadHunter_data_feat_eng['Образование'] = educ_train
HeadHunter_data_feat_eng['Образование'] = HeadHunter_data_feat_eng['Образование'].astype('category')

HeadHunter_data_feat_eng = HeadHunter_data_feat_eng.drop('Образование и ВУЗ', axis=1)
HeadHunter_data_feat_eng['Образование'].unique()
```

* Женщина , 21 год , родилась 13 января 2000
Как вы понимаете, нам необходимо выделить каждый параметр в отдельный столбец.

Создайте два новых признака **"Пол"** и **"Возраст"**. При этом важно учесть:
* Признак пола должен иметь 2 уникальных строковых значения: 'М' - мужчина, 'Ж' - женщина. 
* Признак возраста должен быть представлен целыми числами.

Выполните преобразование, ответьте на контрольные вопросы и удалите признак **"Пол, возраст"** из таблицы.

*Совет: обратите внимание на структуру текста в столбце, в части на то, как разделены параметры пола, возраста и даты рождения между собой - символом ' , '. 
Гарантируется, что структура одинакова для всех строк в таблице. Вы можете воспользоваться этим.*


```python
#ваш код здесь

gender_age_train = HeadHunter_data_feat_eng['Пол, возраст']


def transform_gender(gender_age):
    gender = gender_age.split(',')[0]
    gender = gender.strip()
    if gender == 'Мужчина':
        gender = 'М'
    else:
        gender = 'Ж'
    return gender

def transform_age(gender_age):
    age = gender_age.split(',')[1]
    age = age.strip()
    age = age.split(' ')[0]
    return int(age)

gender_train = gender_age_train.apply(transform_gender)

age_train = gender_age_train.apply(transform_age)

HeadHunter_data_feat_eng['Пол'] = gender_train
HeadHunter_data_feat_eng['Пол'] = HeadHunter_data_feat_eng['Пол'].astype('category')

HeadHunter_data_feat_eng['Возраст'] = age_train

HeadHunter_data_feat_eng = HeadHunter_data_feat_eng.drop('Пол, возраст', axis=1)

HeadHunter_data_feat_eng
```

3. Следующим этапом преобразуем признак **"Опыт работы"**. Его текущий формат - это: **<Опыт работы: n лет m месяцев, периоды работы в различных компаниях…>**. 

Из столбца нам необходимо выделить общий опыт работы соискателя в месяцах, новый признак назовем "Опыт работы (месяц)"

Для начала обсудим условия решения задачи:
* Во-первых, в данном признаке есть пропуски. Условимся, что если мы встречаем пропуск, оставляем его как есть (функция-преобразование возвращает NaN)
* Во-вторых, в данном признаке есть скрытые пропуски. Для некоторых соискателей в столбце стоит значения "Не указано". Их тоже обозначим как NaN (функция-преобразование возвращает NaN)
* В-третьих, нас не интересует информация, которая описывается после указания опыта работы (периоды работы в различных компаниях)
* В-четвертых, у нас есть проблема: опыт работы может быть представлен только в годах или только месяцах. Например, можно встретить следующие варианты:
    * Опыт работы 3 года 2 месяца…
    * Опыт работы 4 года…
    * Опыт работы 11 месяцев…

В результате преобразования у вас должен получиться столбец, содержащий информацию о том, сколько месяцев проработал соискатель.
Выполните преобразование, ответьте на контрольные вопросы и удалите столбец **"Опыт работы"** из таблицы.


```python
#ваш код здесь

expirience_train = HeadHunter_data['Опыт работы']

import re

pattern = re.compile(
    r"опыт работы\s+"
    r"(?:(?P<years>\d+)\s*(?:год|года|лет))?"
    r"\s*"
    r"(?:(?P<months>\d+)\s*(?:месяц|месяца|месяцев|мес))?",
    flags=re.IGNORECASE,
)


def transform_expirience(text: str):
    if text is np.nan or pd.isna(text):
        return np.nan
    if text == "Не указано":
        return np.nan

    match = pattern.search(text)

    if not match:
        return np.nan

    years = match.group("years")
    months = match.group("months")

    years = int(years) if years else 0
    months = int(months) if months else 0

    return years * 12 + months


expirience_train_month = expirience_train.apply(transform_expirience)

HeadHunter_data_feat_eng['Опыт работы (месяц)'] = expirience_train_month

HeadHunter_data_feat_eng = HeadHunter_data_feat_eng.drop('Опыт работы', axis=1)

HeadHunter_data_feat_eng
```

```python
#ваш код здесь
city_relocation_train = HeadHunter_data_feat_eng['Город, переезд, командировки']

# Создаем столбец город
def transform_city(city_relocation):
    if pd.isna(city_relocation):
        return np.nan
    
    million_cities = ['Новосибирск', 'Екатеринбург', 'Нижний Новгород', 'Казань', 'Челябинск', 'Омск', 'Самара', 'Ростов-на-Дону', 'Уфа', 'Красноярск', 'Пермь', 'Воронеж', 'Волгоград' ]
    city_words = city_relocation.split()
    
    city = city_words[0]
    if city in million_cities:
        city = 'город-миллионник'
    elif city not in million_cities and city != 'Москва' and city != 'Санкт-Петербург':
        city = 'другие'
        
    return city

city = city_relocation_train.apply(transform_city)

HeadHunter_data_feat_eng['Город'] = city
HeadHunter_data_feat_eng['Город'].value_counts(normalize=True)

# Создаем столбец Готовность к переезду
pattern = 'не готов к переезду|не готова к переезду'

relocation = ~city_relocation_train.str.contains(pattern, case=False, na=False)

relocation.value_counts()

HeadHunter_data_feat_eng['Готовность к переезду'] = relocation
HeadHunter_data_feat_eng['Готовность к переезду'].value_counts(normalize=True)

# Создаем столбец Готовность к командировкам
pattern1 = 'не готов к командировкам|не готова к командировкам'

work_trip = ~city_relocation_train.str.contains(pattern1, case=False, na=False)



HeadHunter_data_feat_eng['Готовность к командировкам'] = work_trip
HeadHunter_data_feat_eng['Готовность к командировкам'].value_counts(normalize=True)


HeadHunter_data_feat_eng.head()

```

4. Умножить сумму желаемой заработной платы на присоединенный курс валюты (close) и разделить на пропорцию (обратите внимание на пропуски после объединения в этих столбцах), результат занести в новый столбец "ЗП (руб)".


```python
#ваш код здесь
# Загружаем таблицу с курсами валют, удаляем лишние столбцы, преобразуем столбец date
exchange_rates = pd.read_csv('C:\IDE\python_homework\Пайтон для анализа данных\Сессионное задание\ExchangeRates.csv')
exchange_rates = exchange_rates.drop(['per', 'time', 'vol'], axis=1)
exchange_rates['date'] = pd.to_datetime(exchange_rates['date'], dayfirst=True)

# Выделяем дату из столбца Обновление резюме
update_resume_date = pd.to_datetime(HeadHunter_data_feat_eng['Обновление резюме'], dayfirst=True)
update_resume_date = update_resume_date.dt.date
HeadHunter_data_feat_eng['Обновление резюме'] = update_resume_date

# Выделяем значение ЗП и валюту, приводим валюту к ISO
salary_train = HeadHunter_data_feat_eng['ЗП']

currency_name_dict = {'грн.':'UAH', 'USD':'USD', 'EUR':'EUR', 'бел.руб.':'BYN', 'KGS':'KGS', 'сум':'UZS', 'AZN':'AZN', 'KZT':'KZT', 'руб.':'руб'}
   
salary = salary_train.apply(lambda x: x.split()[0])

currency = salary_train.apply(lambda x: x.split()[1])

def transform_currency(currency):
    if currency in currency_name_dict:
        currency_ISO = currency_name_dict[currency]
        return currency_ISO
    
currency_ISO = currency.apply(transform_currency)

HeadHunter_data_feat_eng['salary'] = salary

HeadHunter_data_feat_eng['currency'] = currency_ISO

# Объединяем таблицы с резюме и курсами валют
HeadHunter_data_feat_eng_train = HeadHunter_data_feat_eng.copy()
HeadHunter_data_feat_eng_train['Обновление резюме'] = pd.to_datetime(HeadHunter_data_feat_eng['Обновление резюме'], dayfirst=True)

merge = pd.merge(
    HeadHunter_data_feat_eng_train, 
    exchange_rates, 
    left_on=['Обновление резюме', 'currency'], 
    right_on=['date', 'currency'], 
    how='left' # или 'left', 'right', 'outer'
)

# Заполняем пропуски, получившиеся при объединении
close = merge['close'].fillna(value=1)
merge['close'] = close

proportion = merge['proportion'].fillna(value=1)
merge['proportion'] = proportion

merge['salary'] = merge['salary'].astype(int)

# Рассчитываем сумму ЗП в рублях
salary_rub = (merge['salary']*merge['close'])/merge['proportion']


merge['ЗП (руб)'] = salary_rub

merge = merge.drop(['ЗП', 'salary', 'currency', 'date', 'close', 'proportion'], axis=1)
```

```python
HeadHunter_data_feat_eng = merge
HeadHunter_data_feat_eng = HeadHunter_data_feat_eng.drop(['Город, переезд, командировки'], axis=1)
```

```python
HeadHunter_data_feat_eng['Обновление резюме'] = HeadHunter_data_feat_eng['Обновление резюме'].astype(object)
HeadHunter_data_feat_eng['Образование'] = HeadHunter_data_feat_eng['Образование'].astype(object)
HeadHunter_data_feat_eng['Пол'] = HeadHunter_data_feat_eng['Пол'].astype(object)
HeadHunter_data_feat_eng.info()
```

```python

def outliers_z_score_mod(data, feature, log_scale=False, left = 3, right = 4):
    if isinstance(data, str):  # Проверка, если передан путь к файлу
        data = pd.read_csv(data)
    if log_scale:
        x = np.log(data[feature]+1)
    else:
        x = data[feature]
    mu = x.mean()
    sigma = x.std()
    lower_bound = mu - left * sigma
    upper_bound = mu + right * sigma
    outliers = data[(x < lower_bound) | (x > upper_bound)]
    cleaned = data[(x >= lower_bound) & (x <= upper_bound)]
    return outliers, cleaned

outliers, cleaned = outliers_z_score_mod(HeadHunter_data_feat_eng, feature='Возраст', log_scale=True, left = 3, right = 4)


outliers.index
display(HeadHunter_data_feat_eng['Возраст'].loc[[31137, 32950, 33654]])
HeadHunter_data_feat_eng = HeadHunter_data_feat_eng.drop(outliers.index)
```
