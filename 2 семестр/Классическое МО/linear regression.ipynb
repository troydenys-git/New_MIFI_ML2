```python
import numpy as np

def linear_regression_analytical(X, y):
    # Добавляем столбец единичных значений для смещения (bias)
    X_b = np.c_[np.ones((X.shape[0], 1)), X]  
    
    # Вычисление коэффициентов по аналитической формуле
    # w = (X^T X)^-1 X^T y
    theta = np.linalg.inv(X_b.T.dot(X_b)).dot(X_b.T).dot(y)
    
    return theta

# Пример данных
X = np.array([[1, 2], [2, 5], [3, 7], [4, 10]])  # Матрица признаков
y = np.array([5, 8, 10, 15])  # Целевая переменная

# Нахождение коэффициентов (включая смещение)
theta = linear_regression_analytical(X, y)

print("Коэффициенты модели (включая смещение):")
print(theta)
```

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
class LinearRegressionManual() :

    def __init__(self, learning_rate, iterations) :
        # не забудьте сохранить параметры во внутренние переменные
        self.learning_rate = learning_rate
        self.iterations = iterations
        self.weights = None
       

    def fit(self, X, Y) :
        # классический метод для тренировки
        X = np.hstack((np.ones((X.shape[0], 1)), X))
        self.weights = np.zeros((X.shape[1]))
        
       
        for i in range(self.iterations) :
            # здесь код изменения весов, правильнее его вынести в отдельный метод
            self.update_weights(X, Y)
        return self

    def update_weights(self, X, Y) :
            # здесь код для изменения весов
        y_pred = np.dot(X, self.weights)
        gradients = np.dot(X.T, (y_pred - Y)) / Y.shape[0]
        self.weights -= self.learning_rate * gradients
        return self
        

    def predict( self, X ) :
            # а тут предсказание значений
        X = np.hstack((np.ones((X.shape[0], 1)), X))
        return np.dot(X, self.weights)
```

```python
df = pd.read_csv(r'C:\IDE\2 семестр\Классическое МО\data linear regression.csv')

X = df.iloc[:,:-1].values
Y = df.iloc[:,1].values

model = LinearRegressionManual(iterations = 1000, learning_rate = 0.01)
model.fit(X, Y)

Y_pred = model.predict(X)

plt.scatter(X, Y, color = 'blue', label='Исходные данные')
plt.plot(X, Y_pred, color = 'orange', label='Линия регрессии')
plt.title('Зависимость зарплаты от опыта')
plt.xlabel('Число лет опыта')
plt.ylabel('Зарплата')
plt.legend()

plt.show()
```
