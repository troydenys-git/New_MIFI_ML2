```python
'''Теперь напишем скелеты функции, необходимые для обучения алгоритма логистической регрессии:'''
import numpy as np
def log_loss(w, b, X, y):
    # здесь должны быть рассчитаны параметры градиента и значение функции потерь
    
    z = X@w + b
    epsilon = 1e-15
    y_pred = 1 / (1 + np.exp(-z))

     
    loss = -np.mean(y * np.log(y_pred + epsilon) + (1.0 - y) * np.log(1.0 - y_pred + epsilon))
    
    l = X.shape[1]

    grad_dw = (X.T @ (y_pred - y)) / l
    grad_db = np.sum(y_pred - y) / l
    
   
    
    return grad_dw, grad_db, loss


def optimize(w, b, X, y, n_iterations, eta):
    # потери будем записывать в список для отображения в виде графика
    losses = []
    
    for i in range(n_iterations):
        grad_dw, grad_db, loss = log_loss(w, b, X, y)
        # считаем веса
        w = w - eta*grad_dw
        b = b - eta*grad_db
        losses.append(loss)

    return w, b, losses

def predict(w, b, X):
    z = X@w + b
    y_pred = 1 / (1 + np.exp(-z))
    y_predicted = (y_pred > 0.5).astype(int)
    # За порог отнесения к тому или иному классу примем вероятность 0.5

    return y_predicted

```

```python
X = np.array([[0], [1]])
y = np.array([0, 1])

w = np.zeros(X.shape[1])
b = 0.0

w, b, _ = optimize(w, b, X, y, n_iterations=5000, eta=0.5)
y_pred = predict(w, b, X)

print(*y_pred)
```
